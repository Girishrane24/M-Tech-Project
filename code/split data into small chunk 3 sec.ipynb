```python
from pydub import AudioSegment
from pydub.utils import make_chunks

for s in range(138,159):
    aa = "C:/Users/giris/OneDrive/Desktop/M tech project/Dataset/wav_file_data/file{0}.wav".format(c)
    myaudio = AudioSegment.from_file(aa, "wav") 
    chunk_length_ms = 3000 # pydub calculates in millisec
    chunks = make_chunks(myaudio, chunk_length_ms) #Make chunks of three sec
    del chunks[-1]

    #Export all of the individual chunks as wav files

    for i, chunk in enumerate(chunks):
        chunk_name = "C:/Users/giris/OneDrive/Desktop/M tech project/Dataset/chunk_file_3sec/file{0}.wav".format(chunk_file)
        print ("exporting", chunk_name)
        chunk.export(chunk_name, format="wav")
        chunk_file = chunk_file +1
    print(s)
    c = c+1
```
