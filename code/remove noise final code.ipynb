# final code for remove noise using spectral subtraction

```python
import IPython
from scipy.io import wavfile
import scipy.signal
import numpy as np
import matplotlib.pyplot as plt
import librosa
%matplotlib inline

import pyroomacoustics

from pydub.utils import make_chunks
```

```python

data, rate = librosa.load('C:/Users/giris/Downloads/FICAseparateY.wav',sr = 16000)
time = np.arange(0,len(data))/rate

#data = data / 32768

a = len(time)
```

```python
fig, ax = plt.subplots()
ax.plot(time,data)
print(type(data))
from IPython.display import Audio
Audio(data,rate=rate)
```

```python
output=pyroomacoustics.denoise.spectral_subtraction.apply_spectral_sub(data,
                                                                       nfft=512,
                                                                       db_reduc=25,
                                                                       lookback=12,
                                                                       beta=30, 
                                                                       alpha=4)
```

```python
fig, ax = plt.subplots()
ax.plot(time,output)
print(type(output))

from IPython.display import Audio
Audio(output,rate=rate)
```
