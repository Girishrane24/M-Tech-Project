```python

audio, sfreq = librosa.load('C:/Users/giris/Downloads/XC291448 - Red-whiskered Bulbul - Pycnonotus jocosus.wav',sr = 16000)
time = np.arange(0,len(audio))/sfreq

print()
print(len(audio))
print()
print(time)
print()
a = len(time)
print()
print(sfreq)
print("Time duration : {}",a/sfreq)


fig, ax = plt.subplots()
ax.plot(time,audio)
print(type(audio))
from IPython.display import Audio
Audio(audio,rate=sfreq)
```

```python
#use STFT
S = np.abs(librosa.stft(audio))
S
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
img = librosa.display.specshow(S, cmap='gray_r', y_axis='log', x_axis='time', ax=ax)
ax.set(title='Inverted grayscale')
fig.colorbar(img, ax=ax, format="%+2.f dB")
```
