```python
data=pd.read_csv("C:/Users/giris/OneDrive/Desktop/dataset for m tech project_new/3 sec chunk file for Mtech data set.csv")
```

```python
data
```

```python

import scipy.io.wavfile
from spafe.utils import vis
from spafe.features.gfcc import gfcc
import librosa

# init input vars
num_ceps = 13
low_freq = 0
high_freq = 8000
nfilts = 40
nfft = 512
dct_type = 2
use_energy = False
lifter = 2


x_train_1=[]

x_train_=[]
y_train_=[]

path="C:/Users/giris/OneDrive/Desktop/dataset for m tech project_new/3_sec_chunk_final"
for i in tqdm(range(len(data))):
    #fold_no=str(data.iloc[i]["fold"])
    file=data.iloc[i]["file_name"]
    label=data.iloc[i]["label"]
    filename=path+"/"+file
    
       
    #read wave file
    y,sr=librosa.load(filename,sr = 16000)
    
    gfccs = gfcc(sig=y,
             fs=sr,
             num_ceps=num_ceps,
             nfilts=nfilts,
             nfft=nfft,
             low_freq=low_freq,
             high_freq=high_freq,
             win_len = 0.020,
             win_hop = 0.010,
             dct_type=dct_type,
             use_energy=use_energy,
             lifter=lifter,
                        normalize = False)

    x_train_1.append(gfccs)
    y_train_.append(label)
    
    
```

```python
x_train_1 = np.asarray(x_train_1)
print(x_train_1.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x_train_1,y_train_,test_size=0.30, random_state=1)
print(X_train.shape)
print(X_test.shape)
```

```python
from keras.utils import to_categorical
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
print(y_train.shape)
print(y_test.shape)
print(y_train[2])
```

```python

dim_1 = x_train_1.shape[1]
dim_2 = x_train_1.shape[2]
print(dim_1)
print(dim_2)
```

```python
###########        train set       :::::::::::::::::::::
channel_dim = 1
channels = 1
X_train = X_train.reshape((X_train.shape[0], dim_1, dim_2, channels))
X_test = X_test.reshape((X_test.shape[0], dim_1, dim_2, channels))  

input_shape = (dim_1, dim_2, channels)

print(X_train.shape)
print(X_test.shape)
```

```python
model = Sequential()

model.add(Conv2D(16, kernel_size=(3, 3), activation='relu', input_shape=input_shape))
model.add(MaxPooling2D(pool_size=(2, 2)))


model.add(Conv2D(32, kernel_size=(3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))


model.add(Flatten())

model.add(Dense(1024, activation='relu'))
#model.add(Dropout(0.2))
model.add(Dense(10,activation = 'softmax'))
model.summary()
```

```python
model.compile(loss=tensorflow.keras.losses.categorical_crossentropy,
              optimizer=tensorflow.keras.optimizers.Adam(learning_rate=0.001),
              metrics=['accuracy'])
```

```python

history=model.fit(X_train, y_train,validation_data = (X_test, y_test) ,batch_size=16,epochs=30,shuffle=True,verbose=2)

```

```python
#import library
import numpy as np
import tensorflow as tf
from tensorflow import keras
from matplotlib import pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPooling2D
from sklearn.model_selection import train_test_split
import tensorflow.keras
```

```python
data=pd.read_csv("C:/Users/giris/OneDrive/Desktop/dataset for m tech project_new/3 sec chunk file for Mtech data set.csv")
```

```python
data
```

```python

import scipy.io.wavfile
from spafe.utils import vis
from spafe.features.gfcc import gfcc
import librosa

pad_width = 320
max_pad_len=320
# init input vars
num_ceps = 13
low_freq = 0
high_freq = 8000
nfilts = 30
nfft = 512
dct_type = 2
use_energy = False
lifter = 2

x_train_1=[]

x_train_=[]
y_train_=[]

path="C:/Users/giris/OneDrive/Desktop/dataset for m tech project_new/3_sec_chunk_final"
for i in tqdm(range(len(data))):
    #fold_no=str(data.iloc[i]["fold"])
    file=data.iloc[i]["file_name"]
    label=data.iloc[i]["label"]
    filename=path+"/"+file
    
       
    #read wave file
    y,sr=librosa.load(filename,sr = 16000)
    #y,sr = wavfile.read(filename)
    #features_mfcc=mfcc(y, sr)
    #features_mfcc = features_mfcc.T
    #mfccs = np.mean((features_mfcc),axis = 0)
    
    gfccs = gfcc(sig=y,
             fs=sr,
             num_ceps=num_ceps,
             nfilts=nfilts,
             nfft=nfft,
             low_freq=low_freq,
             high_freq=high_freq,
             win_len = 0.020,
             win_hop = 0.010,
             dct_type=dct_type,
             use_energy=use_energy,
             lifter=lifter,
                        normalize = False)
    
    deltas = librosa.feature.delta(gfccs, width=9, order=1,axis=1)
    deltas_deltas = librosa.feature.delta(gfccs, width=9, order=2,axis=1)
    combined = np.hstack((gfccs, deltas,deltas_deltas))
    ####
    #delete some coefficients  
    combined = np.delete(combined, [3,7,11,26,27,28,29,30,34,35,37,38], 1)
    
    
    
    x_train_1.append(combined)
    
    y_train_.append(label)
```
