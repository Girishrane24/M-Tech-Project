# Feature Selection using ANN

```python
import numpy as np
import pandas as pd
import keras
from keras.utils import to_categorical
from keras import optimizers
```

```python
data = pd.read_csv("C:/Users/giris/OneDrive/Desktop/dataset for m tech project_new/MFCC/feature_select_for_knn.csv")
```

```python
data.head()
```

```python
X = data.iloc[:, :-1].values
y = data.iloc[:, 27].values
print(X)

print(y)
```

```python
#get training and testing dataset
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30)



print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from keras.utils import to_categorical
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

```python
#import classes
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Dense,Dropout
```

```python
#model definition:
model = Sequential()
model.add(Dense(32,activation='relu',input_dim = 27))
model.add(Dense(64,activation='relu'))
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation = 'softmax'))

model.summary()
```

```python
model.compile(loss = 'categorical_crossentropy',
              optimizer = 'adam',
             metrics =['accuracy'])
```

```python
score_1=model.fit(X_train,y_train,
          validation_data = (X_test, y_test),
          epochs = 50,
          batch_size  =16,verbose = 2)
```

```python
score = model.evaluate(X_test,y_test)
```

```python
score
```

```python
#print(acc * 100)

plt.plot(score_1.history['accuracy'])
plt.plot(score_1.history['val_accuracy'])
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title('ANN model Accuracy')
plt.legend(['train Accuracy', 'validation Accuracy'], loc='lower right')
plt.show()

```

```python

plt.plot(score_1.history['loss'])
plt.plot(score_1.history['val_loss'])
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title('ANN Model Loss')
plt.legend(['train Loss', 'validation Loss'], loc='upper right')
plt.show()

```
